## 🔴 AI Groundwater Risk Intelligence

This section focuses on analyzing groundwater risk using different indicators.  
It helps identify high-risk areas and understand the factors affecting groundwater depletion.

##### Import Libraries

```python
import pandas as pd
import plotly.express as px
```

##### Load Dataset

```python
df = pd.read_csv("groundwater_ml_dataset_cleaned.csv")
df.head()
```

 ##### 🎨 Color Theme Used

White background with Plasma color scale:
- Light → Low Risk  
- Dark → High Risk  
- Clean & professional dashboard look


### 📊 1. Risk Score Distribution

This graph shows how risk scores are spread across all regions.

```python
fig1 = px.histogram(
    df,
    x='risk_score',
    nbins=25,
    title='Risk Score Distribution',
    color_discrete_sequence=px.colors.sequential.Blugrn_r
)

fig1.update_layout(template='plotly_white', title_x=0.5)
fig1.show()
```

### 📊 2. Risk Score by Category

This graph compares average risk across different categories.

```python
avg_risk = df.groupby('category')['risk_score'].mean().reset_index()

fig2 = px.bar(
    avg_risk,
    x='category',
    y='risk_score',
    color='risk_score',
    color_continuous_scale=px.colors.sequential.Blugrn,
    title='Risk Score by Category'
)

fig2.update_layout(template='plotly_white', title_x=0.5)
fig2.show()
```

### 📊3. Risk Score vs Stress Level

Shows relationship between stress level and groundwater risk.

```python
fig3 = px.scatter(
    df,
    x='stress_level',
    y='risk_score',
    color='risk_score',
    size='risk_score',
    color_continuous_scale=px.colors.sequential.Blugrn,
    title='Risk Score vs Stress Level'
)

fig3.update_layout(template='plotly_white', title_x=0.5)
fig3.show()
```

### 📊 4. Risk Score vs Extraction Ratio

Shows how extraction impacts groundwater risk.

```python
fig4 = px.box(
    df,
    x='category',
    y='extraction_ratio',
    color='category',
    title='Extraction Ratio Distribution by Category',
    color_discrete_sequence=px.colors.sequential.Blugrn_r
)

fig4.update_layout(
    template='plotly_white',
    title_x=0.5
)

fig4.show()
```

### 📊 5. High vs Low Risk Comparison

Classifies regions into high and low risk categories.

```python
df['risk_level'] = df['risk_score'].apply(
    lambda x: 'High Risk' if x > df['risk_score'].mean() else 'Low Risk'
)

risk_count = df['risk_level'].value_counts().reset_index()
risk_count.columns = ['risk_level', 'count']

fig5 = px.pie(
    risk_count,
    names='risk_level',
    values='count',
    title='High vs Low Risk Areas',
    color='count',
    color_discrete_sequence=px.colors.sequential.Blugrn_r
)

fig5.update_layout(template='plotly_white', title_x=0.5)
fig5.show()
```

### 📊 6 Risk Score by Category

- This chart shows the average risk score across different categories.

- It helps in identifying which category has higher risk levels 
and requires more attention.

- Outer area → Higher risk  
- Inner area → Lower risk

```python
avg_df = df.groupby('category')['risk_score'].mean().reset_index()

g1 = px.bar(
    avg_df,
    x='category',
    y='risk_score',
    color='category',
    title='Average Risk Score by Category',
    color_discrete_sequence=px.colors.sequential.Blugrn_r
)

g1.update_layout(template='plotly_white', title_x=0.5)
g1.show()
```

### 📊 7 Top 10 High Risk Districts

- This chart highlights districts with the highest risk scores.

- It helps quickly identify critical regions that require 
immediate attention or intervention.

```python
top_districts = df.sort_values('risk_score', ascending=False).head(10)

fig2 = px.treemap(
    top_districts,
    path=['district'],
    values='risk_score',
    color='risk_score',
    color_continuous_scale=px.colors.sequential.Blugrn,
    title='Top 10 High Risk Districts'
)

fig2.update_layout(title_x=0.5)
fig2.show()
```

### 8 Risk Level Donut Chart
- 💡 Insight:

Shows proportion of safe vs dangerous groundwater zones.

```python
df['risk_band'] = pd.cut(
    df['risk_score'],
    bins=[0, 0.33, 0.66, 1],
    labels=['Low', 'Medium', 'High']
)

donut = df['risk_band'].value_counts().reset_index()
donut.columns = ['risk_band', 'count']

fig = px.pie(
    donut,
    names='risk_band',
    values='count',
    hole=0.6,
    title='Risk Level Distribution (Donut Chart)',
    color_discrete_sequence=px.colors.sequential.Blugrn
)

fig.update_layout(template='plotly_white', title_x=0.5)
fig.show()
```

### 9. Treemap: High Risk District Clusters
- 💡 Insight:

Quickly identifies problem districts inside states.

```python
district_tree = df.groupby(['state', 'district'])['risk_score'].mean().reset_index()

fig = px.treemap(
    district_tree,
    path=['state', 'district'],
    values='risk_score',
    color='risk_score',
    color_continuous_scale='Tealgrn',
    title='Treemap: District-wise Risk Structure'
)

fig.update_layout(template='plotly_white', title_x=0.5)
fig.show()
```
