```python
import camelot
import pandas as pd

# PDF file path
pdf_path = "groundwater.pdf"

# Extract tables from all pages
tables = camelot.read_pdf(pdf_path, pages="all", flavor="stream")

print("Total tables found:", tables.n)

# Store tables
dfs = []

for i, table in enumerate(tables):
    df = table.df
    dfs.append(df)

# Combine all tables
combined_df = pd.concat(dfs, ignore_index=True)

print("Raw data shape:", combined_df.shape)

print(combined_df.head())
```

```python
print("Total Columns:", combined_df.shape[1])
print(combined_df.columns)
```

```python
combined_df = combined_df.replace("", pd.NA)

combined_df = combined_df.dropna(axis=1, how="all")

print(combined_df.shape)
```

```python
df = combined_df.iloc[:, :6]   # first useful columns
```

```python
print(combined_df.columns)
```

```python
df.columns = [
    "state",
    "district",
    "annual_recharge",
    "extractable_resource",
    "annual_extraction",
    "stage_of_development"
]
```

```python
def classify(stage):
    try:
        stage = float(stage)
        if stage < 70:
            return "Safe"
        elif stage < 90:
            return "Semi Critical"
        elif stage < 100:
            return "Critical"
        else:
            return "Over Exploited"
    except:
        return None

df["category"] = df["stage_of_development"].apply(classify)
```

```python
df = df.dropna()

print(f"Rows after removing null values: {df.shape[0]}")
print(f"Null values remaining: {df.isnull().sum().sum()}")
```

```python
# ================================
# STEP 1: Convert numeric columns
# ================================
numeric_cols = [
    "annual_recharge",
    "extractable_resource",
    "annual_extraction",
    "stage_of_development"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# ================================
# STEP 2: Drop null values again (after conversion)
# ================================
df = df.dropna().reset_index(drop=True)

# ================================
# STEP 3: Add NEW ML-USEFUL FEATURES
# ================================

# 1️⃣ Extraction Ratio
df["extraction_ratio"] = df["annual_extraction"] / df["annual_recharge"]

# 2️⃣ Utilization Rate
df["utilization_rate"] = df["annual_extraction"] / df["extractable_resource"]

# 3️⃣ Stress Level (numeric form of stage)
df["stress_level"] = df["stage_of_development"] / 100

# 4️⃣ Risk Score (important ML feature)
df["risk_score"] = (
    df["extraction_ratio"] * 0.4 +
    df["utilization_rate"] * 0.4 +
    df["stress_level"] * 0.2
)

# ================================
# STEP 4: Final Cleaning
# ================================
df = df.replace([float("inf"), -float("inf")], pd.NA)
df = df.dropna().reset_index(drop=True)

# ================================
# STEP 5: Add Year Column
# ================================
df["year"] = 2024

# ================================
# STEP 6: Final Dataset Check
# ================================
print("\nFinal Dataset Info:\n")
print(df.info())

print("\nSample Data:\n")
print(df.head())

print("\nNo. of Rows:", df.shape[0])
print("No. of Columns:", df.shape[1])

# ================================
# STEP 7: Save Dataset
# ================================
df.to_csv("groundwater_ml_dataset.csv", index=False)

print("\n✅ CLEAN ML DATASET CREATED SUCCESSFULLY")
```
